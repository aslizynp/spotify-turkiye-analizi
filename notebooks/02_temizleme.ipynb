```python
import pandas as pd
import sqlite3

df = pd.read_csv("../data/universal_top_spotify_songs.csv")
print("Başlangıç:", df.shape)
```

```python
df["country"] = df["country"].fillna("GLOBAL")
print("GLOBAL satır sayısı:", (df["country"] == "GLOBAL").sum())
```

```python
df["snapshot_date"] = pd.to_datetime(df["snapshot_date"])
df["album_release_date"] = pd.to_datetime(df["album_release_date"], errors="coerce")
print(df[["snapshot_date", "album_release_date"]].dtypes)
```

```python
once = len(df)
df = df.dropna(subset=["name", "artists"])
print("Silinen satır:", once - len(df))
```

```python
df_tr = df[df["country"].isin(["TR", "GLOBAL"])].copy()
print(df_tr.shape)
print(df_tr["country"].value_counts())
```

```python
df_tr.to_csv("../data/spotify_tr_global.csv", index=False)

conn = sqlite3.connect("../data/spotify.db")
df_tr.to_sql("songs", conn, if_exists="replace", index=False)

pd.read_sql("SELECT country, COUNT(*) AS satir FROM songs GROUP BY country", conn)
```
